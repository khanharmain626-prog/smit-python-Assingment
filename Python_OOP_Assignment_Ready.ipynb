## Variables in Classes and Multiple Inheritance

**Topics:** Local variables, instance variables, class variables, and multiple inheritance.

## 1. Local Variable

A local variable is created inside a method or function. It can normally be used only inside that method or function.

```python
class Student:
    def show(self):
        name = "Usman"  # Local variable
        print(name)

s1 = Student()
s1.show()
```

## 2. Instance Variables

Instance variables belong to a particular object. In this example, each student object can have its own `name` and `age`.

```python
class Student:
    def __init__(self, name, age):
        self.name = name  # Instance variable
        self.age = age   # Instance variable

    def show_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

student1 = Student("Usman", 20)
student2 = Student("Ali", 19)

student1.show_info()
student2.show_info()
```

## 3. Class Variable

A class variable is shared by objects of the class unless an object overrides it. The `company` variable below belongs to the `Teacher` class.

```python
class Teacher:
    school = "Saylani"  # Class variable

    def __init__(self, subject, timing):
        self.subject = subject  # Instance variable
        self.timing = timing    # Instance variable

    def show_info(self):
        print(f"School: {self.school}")
        print(f"Subject: {self.subject}")
        print(f"Timing: {self.timing}")

teacher1 = Teacher("Math", "10 PM - 1 AM")
print(teacher1.school)
teacher1.show_info()
```

## 4. Multiple Inheritance

Multiple inheritance means that one child class inherits from two or more parent classes.

```python
class Father:
    def father_skill(self):
        print("Father: Business")

class Mother:
    def mother_skill(self):
        print("Mother: Cooking")

class Child(Father, Mother):
    def child_skill(self):
        print("Child: Coding")

child1 = Child()
child1.father_skill()
child1.mother_skill()
child1.child_skill()
```

## 5. Real-World Example: AI Employee

`AIEmployee` inherits methods from both `Employee` and `AIEngineer`. It can use methods from both parent classes.

```python
class Employee:
    company = "Saylani"

    def employee_info(self):
        print("I am an employee.")

    def show_company(self):
        print(f"Company: {self.company}")

class AIEngineer:
    def ai_work(self):
        print("I build AI models.")

    def train_model(self):
        print("I train machine-learning models.")

class AIEmployee(Employee, AIEngineer):
    def __init__(self, name, age):
        self.name = name  # Instance variable
        self.age = age    # Instance variable

    def show_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Company: {self.company}")

ai_employee = AIEmployee("Usman", 20)
ai_employee.show_info()
ai_employee.employee_info()
ai_employee.show_company()
ai_employee.ai_work()
ai_employee.train_model()
```

## Conclusion

- **Local variable:** Defined inside a function or method.
- **Instance variable:** Stores data for an individual object, commonly using `self`.
- **Class variable:** Shared at the class level.
- **Multiple inheritance:** A child class inherits features from more than one parent class.
